```python
#import frameworks
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
#Load Data Sets from from Kaggle
df = pd.read_csv('Salary_Data.csv')
```

```python
#Inspect the data
#Shape
print(df.shape)
```

```python
#Data Types
print( df.dtypes)
```

```python
#Missing Values
print(df.isnull().sum())
```

```python
#Identifying and Separating The Variables
x = df[['YearsExperience']]
y = df['Salary']
print("x shape:", x.shape)
print("y shape:", y.shape)
```

```python
#Splitting The Data
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
#Train the Model
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
#Prediction on the Set
y_predict=model.predict(x_test)
print(y_predict)
```

```python
#Evaluate the model using MSE
mse=mean_squared_error(y_test,y_predict)
print(mse)
```

```python
#On average, your model's predictions deviate from actual values by approximately ±$7,059
```

```python
#Ploting it into a Graph(Visualize the regression Line)
import matplotlib.pyplot as plt
plt.scatter(x,y,color="blue", label="Actual Data")
plt.plot(x,model.predict(x),color="orange",linewidth=2,label="Regression Line")
plt.xlabel("Years of Experince")
plt.ylabel("Salary")
plt.title("Salary Prediction based on Years of Experince")
plt.legend()
plt.show()
```
